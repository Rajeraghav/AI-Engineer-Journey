```python
# ============================================================
# CNN IMPLEMENTATION FOR CUSTOM MNIST CSV DATASET
# ============================================================

# ------------------------------------------------------------
# 1. Import Libraries
# ------------------------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score
)

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Conv2D,
    MaxPooling2D,
    Flatten,
    Dense,
    Dropout
)


# ------------------------------------------------------------
# 2. Upload Dataset
# ------------------------------------------------------------

uploaded = files.upload()

# Get uploaded filename automatically
filename = list(uploaded.keys())[0]

print("Uploaded file:", filename)


# ------------------------------------------------------------
# 3. Load Dataset
# ------------------------------------------------------------

df = pd.read_csv(filename)

print("\nDataset Shape:")
print(df.shape)

print("\nFirst 5 Rows:")
print(df.head())


# ------------------------------------------------------------
# 4. Check Dataset Information
# ------------------------------------------------------------

print("\nDataset Information:")
print(df.info())

print("\nColumn Names:")
print(df.columns.tolist())


# ------------------------------------------------------------
# 5. Separate Features and Target
# ------------------------------------------------------------

# c0 to c63 = pixel features
# target = class label

X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values

print("\nFeature Shape:")
print(X.shape)

print("\nTarget Shape:")
print(y.shape)


# ------------------------------------------------------------
# 6. Display Unique Classes
# ------------------------------------------------------------

print("\nUnique Classes:")
print(np.unique(y))

print("\nNumber of Classes:")
print(len(np.unique(y)))


# ------------------------------------------------------------
# 7. Convert Data Type
# ------------------------------------------------------------

X = X.astype("float32")
y = y.astype("int32")


# ------------------------------------------------------------
# 8. Normalize Pixel Values
# ------------------------------------------------------------

# Your dataset contains pixel values from 0 to 16
X = X / 16.0

print("\nMinimum Pixel Value:", X.min())
print("Maximum Pixel Value:", X.max())


# ------------------------------------------------------------
# 9. Reshape Data
# ------------------------------------------------------------

# 64 pixels = 8 x 8
# 1 = grayscale channel

X = X.reshape(-1, 8, 8, 1)

print("\nShape After Reshaping:")
print(X.shape)


# ------------------------------------------------------------
# 10. Display Sample Image
# ------------------------------------------------------------

plt.figure(figsize=(4, 4))

plt.imshow(
    X[0].reshape(8, 8),
    cmap="gray"
)

plt.title("Sample Image - Label: " + str(y[0]))
plt.axis("off")
plt.show()


# ------------------------------------------------------------
# 11. Train-Test Split
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining Data Shape:")
print(X_train.shape)

print("\nTesting Data Shape:")
print(X_test.shape)

print("\nTraining Labels Shape:")
print(y_train.shape)

print("\nTesting Labels Shape:")
print(y_test.shape)


# ------------------------------------------------------------
# 12. Build CNN Model
# ------------------------------------------------------------

model = Sequential([

    # First Convolution Layer
    Conv2D(
        filters=32,
        kernel_size=(3, 3),
        activation="relu",
        input_shape=(8, 8, 1)
    ),

    # First Pooling Layer
    MaxPooling2D(
        pool_size=(2, 2)
    ),

    # Second Convolution Layer
    Conv2D(
        filters=64,
        kernel_size=(3, 3),
        activation="relu"
    ),

    # Flatten
    Flatten(),

    # Fully Connected Layer
    Dense(
        128,
        activation="relu"
    ),

    # Dropout
    Dropout(0.5),

    # Output Layer
    Dense(
        10,
        activation="softmax"
    )
])


# ------------------------------------------------------------
# 13. Display Model Architecture
# ------------------------------------------------------------

print("\nCNN Model Summary:")
model.summary()


# ------------------------------------------------------------
# 14. Compile CNN
# ------------------------------------------------------------

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


# ------------------------------------------------------------
# 15. Train CNN
# ------------------------------------------------------------

history = model.fit(
    X_train,
    y_train,
    epochs=20,
    batch_size=32,
    validation_split=0.10,
    verbose=1
)


# ------------------------------------------------------------
# 16. Evaluate CNN
# ------------------------------------------------------------

test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test,
    verbose=0
)

print("\n====================================")
print("CNN MODEL PERFORMANCE")
print("====================================")

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)


# ------------------------------------------------------------
# 17. Make Predictions
# ------------------------------------------------------------

y_pred_probability = model.predict(
    X_test,
    verbose=0
)

# Convert probabilities to class labels
y_pred = np.argmax(
    y_pred_probability,
    axis=1
)


# ------------------------------------------------------------
# 18. Calculate Accuracy
# ------------------------------------------------------------

accuracy = accuracy_score(
    y_test,
    y_pred
)

print("\nPrediction Accuracy:", accuracy)


# ------------------------------------------------------------
# 19. Classification Report
# ------------------------------------------------------------

print("\n====================================")
print("CLASSIFICATION REPORT")
print("====================================")

print(
    classification_report(
        y_test,
        y_pred
    )
)


# ------------------------------------------------------------
# 20. Confusion Matrix
# ------------------------------------------------------------

cm = confusion_matrix(
    y_test,
    y_pred
)

print("\n====================================")
print("CONFUSION MATRIX")
print("====================================")

print(cm)


# ------------------------------------------------------------
# 21. Display Confusion Matrix
# ------------------------------------------------------------

plt.figure(figsize=(8, 6))

plt.imshow(cm)

plt.title("CNN Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")

plt.colorbar()

plt.xticks(
    np.arange(10),
    np.arange(10)
)

plt.yticks(
    np.arange(10),
    np.arange(10)
)

# Display values inside matrix
for i in range(10):
    for j in range(10):
        plt.text(
            j,
            i,
            cm[i, j],
            ha="center",
            va="center"
        )

plt.show()


# ------------------------------------------------------------
# 22. Plot Training Accuracy
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("CNN Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()

plt.show()


# ------------------------------------------------------------
# 23. Plot Training Loss
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("CNN Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()

plt.show()


# ------------------------------------------------------------
# 24. Predict One Image
# ------------------------------------------------------------

index = 0

sample_image = X_test[index]

sample_input = np.expand_dims(
    sample_image,
    axis=0
)

prediction = model.predict(
    sample_input,
    verbose=0
)

predicted_class = np.argmax(
    prediction
)

actual_class = y_test[index]

print("\n====================================")
print("SINGLE IMAGE PREDICTION")
print("====================================")

print("Actual Digit:", actual_class)
print("Predicted Digit:", predicted_class)


# ------------------------------------------------------------
# 25. Display Predicted Image
# ------------------------------------------------------------

plt.figure(figsize=(4, 4))

plt.imshow(
    sample_image.reshape(8, 8),
    cmap="gray"
)

plt.title(
    "Actual: " + str(actual_class) +
    " | Predicted: " + str(predicted_class)
)

plt.axis("off")

plt.show()


# ------------------------------------------------------------
# 26. Display First 10 Predictions
# ------------------------------------------------------------

print("\n====================================")
print("FIRST 10 PREDICTIONS")
print("====================================")

for i in range(10):

    print(
        "Image", i + 1,
        "| Actual:", y_test[i],
        "| Predicted:", y_pred[i]
    )


# ------------------------------------------------------------
# 27. Save CNN Model
# ------------------------------------------------------------

model.save("MNIST_CNN_Model.keras")

print("\nCNN model saved successfully as MNIST_CNN_Model.keras")
```
